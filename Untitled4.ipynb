```python
# Project 1: Handwritten Digit Recognizer using CNN

import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras import layers, models

# Step 1: Load MNIST dataset automatically
(x_train, y_train), (x_test, y_test) = (
    tf.keras.datasets.mnist.load_data()
)

# Step 2: Normalize image values
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# Step 3: Build CNN model
model = models.Sequential([
    layers.Input(shape=(28, 28, 1)),
    layers.Reshape((28, 28, 1)),
    layers.Conv2D(32, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(64, activation="relu"),
    layers.Dense(10, activation="softmax")
])

# Step 4: Compile model
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# Step 5: Train the model
model.fit(
    x_train, y_train,
    epochs=3,
    batch_size=128,
    validation_split=0.1
)

# Step 6: Evaluate model
loss, accuracy = model.evaluate(x_test, y_test, verbose=0)
print("Test Accuracy:", round(accuracy * 100, 2), "%")

# Step 7: Predict a handwritten digit
index = 0
prediction = model.predict(x_test[index:index+1], verbose=0)
predicted_digit = np.argmax(prediction)

plt.imshow(x_test[index], cmap="gray")
plt.title("Predicted: " + str(predicted_digit))
plt.axis("off")
plt.show()

print("Actual Digit:", y_test[index])
print("Predicted Digit:", predicted_digit)

# Step 8: Save trained model
model.save("handwritten_digit_model.keras")
print("Model saved successfully!")
```
